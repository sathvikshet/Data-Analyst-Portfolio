```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
df = pd.read_csv(r"C:\analytics portfolio\machine-learning-projects\customer-churn-prediction\data\Telcom-Customer-Churn.csv")
df.head()
```

```python
sns.countplot(x="Churn", data=df)

plt.title("Customer Churn Distribution")
plt.show()
```

```python
sns.boxplot(x="Churn", y="MonthlyCharges", data=df)

plt.title("Monthly Charges vs Churn")
plt.show()
```

```python
sns.countplot(x="Contract", hue="Churn", data=df)

plt.title("Contract Type vs Churn")
plt.xticks(rotation=30)
plt.show()
```

```python
df["Churn"] = df["Churn"].map({"Yes":1,"No":0})
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

plt.figure(figsize=(12,6))

sns.heatmap(df.corr(numeric_only=True), cmap="coolwarm", annot=True)

plt.title("Feature Correlation Heatmap")

plt.show()
```

```python
plt.savefig("../visuals/churn_distribution.png")
```

EDA Insights

1. Customers with shorter tenure churn more frequently.

2. Month-to-month contracts have the highest churn rate.

3. Higher monthly charges are associated with higher churn probability.
